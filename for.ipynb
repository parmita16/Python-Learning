```python
for i in range(5):
    print("* " * 5)
```

```python
for i in range(1, 6):
    print("* " * i)
```

```python
for i in range(5, 0, -1):
    print("* " * i)
```

```python
rows = 5

for i in range(rows):
    print(" " * (rows - i - 1) + "* " * (i + 1))
```

```python
rows = 5

for i in range(rows, 0, -1):
    print(" " * (rows - i) + "* " * i)
```

```python
rows = 5

for i in range(1, rows + 1):
    print(" " * (rows - i) + "* " * i)

for i in range(rows - 1, 0, -1):
    print(" " * (rows - i) + "* " * i)
```

```python
n = 5

for i in range(n):
    for j in range(n):
        if i == 0 or i == n - 1 or j == 0 or j == n - 1:
            print("*", end=" ")
        else:
            print(" ", end=" ")
    print()
```
